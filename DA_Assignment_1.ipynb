```python
Price_USD=df['Price_USD']
m1=Price_USD.mean()
m2=Price_USD.median()
m3=Price_USD.mode()
print(m1)
print(m2)
print(m3)
```

```python
q3=Price_USD.quantile(0.75)
q1=Price_USD.quantile(0.25)
iqr=q3-q1
print(iqr)
```

```python
range=Price_USD.max()-Price_USD.min()
print(range)
```

```python
var=Price_USD.var()
std=Price_USD.std()
print(var)
print(std)
```

```python
iqr_value=Price_USD.quantile(0.75)-Price_USD.quantile(0.25)
print(iqr_value)
```
