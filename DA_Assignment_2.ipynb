```python
import pandas as pd
dataset_1=pd.read_csv('/content/car (1) (1) - car (1) (1) (1).csv')
```

```python
dataset_2=pd.read_csv('/content/pokemon (1) - pokemon (1).csv')
```

### 1. Most expensive car for each car company make.

```python
most_expensive_by_make = dataset_1.loc[dataset_1.groupby('Make')['MSRP'].idxmax()]
display(most_expensive_by_make[['Make', 'Model', 'MSRP']])
```

### 2. Most expensive convertible vehicle for each company.

```python
convertibles = dataset_1[dataset_1['Vehicle Style'].str.contains('Convertible', case=False, na=False)]
most_expensive_convertible_by_make = convertibles.loc[convertibles.groupby('Make')['MSRP'].idxmax()]
display(most_expensive_convertible_by_make[['Make', 'Model', 'Vehicle Style', 'MSRP']])
```

### 3. List all the cars which are automatic.

```python
automatic_cars = dataset_1[dataset_1['Transmission Type'] == 'AUTOMATIC']
display(automatic_cars[['Make', 'Model', 'Transmission Type']].head())
```

### 4. Premium cars of each car company

```python
# This task is a bit ambiguous, assuming 'Premium' refers to 'Luxury' or 'Performance' in 'Market Category'.
premium_cars = dataset_1[dataset_1['Market Category'].str.contains('Luxury|Performance', case=False, na=False)]
display(premium_cars[['Make', 'Model', 'Market Category']].head())
```

### 5. Average price of cars each car company make.

```python
average_price_by_make = dataset_1.groupby('Make')['MSRP'].mean().sort_values(ascending=False)
display(average_price_by_make.head())
```

## Pokémon Dataset Analysis

### 1. Check for Missing Values

```python
missing_values = dataset_2.isnull().sum()
print("Missing values in each column:")
display(missing_values[missing_values > 0])
```

### 2. Explore Pokémon Types (type1 and type2)

```python
print("Top 10 most common Primary Types (type1):")
display(dataset_2['type1'].value_counts().head(10))

print("\nTop 10 most common Secondary Types (type2):")
display(dataset_2['type2'].value_counts().head(10))
```

### 3. Descriptive Statistics for Numerical Features

```python
print("Descriptive statistics for numerical columns:")
display(dataset_2.describe())
```

### 4. Distribution of Legendary vs. Non-Legendary Pokémon

```python
legendary_counts = dataset_2['is_legendary'].value_counts()
print("Distribution of Legendary vs. Non-Legendary Pokémon:")
display(legendary_counts)

print("\nPercentage of Legendary Pokémon:")
display(dataset_2['is_legendary'].value_counts(normalize=True) * 100)
```
