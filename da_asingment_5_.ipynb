```python
import pandas as pd

# Basic probabilities (using df consistently to ensure 'Status' column is available)
p_a = (df['Gender'] == 'Male').mean()
p_b = (df['Status'] == 'Pass').mean()
p_c = (df['Attendance'] > 75).mean()  # P(C) from earlier sections

# 1. P(A ∩ B) – Probability that a student is both Male AND Passes
p_a_and_b = ((df['Gender'] == 'Male') & (df['Status'] == 'Pass')).mean()

# 2. P(A ∩ B) verified using Multiplication Rule: P(A ∩ B) = P(A) % P(B|A)
p_b_given_a = df.loc[df['Gender'] == 'Male', 'Status'].eq('Pass').mean()
p_a_and_b_verified = p_a * p_b_given_a

# 3. P(A ∦ B) – Probability that a student is Male OR Passes (or both)
p_a_or_b = p_a + p_b - p_a_and_b

# 4. Complements
p_not_b = 1 - p_b  # Probability that a student does Not Pass
p_attendance_le_75 = 1 - p_c  # Probability that Attendance is <= 75%

# Print results
print(f"P(A ∩ B): {p_a_and_b:.4f}")
print(f"P(A ∦ B): {p_a_or_b:.4f}")
print(f"Multiplication Check P(A) % P(B|A): {p_a_and_b_verified:.4f}")
print(f"P(Not Pass): {p_not_b:.4f}")
print(f"P(Attendance ≤ 75%): {p_attendance_le_75:.4f}")
```
